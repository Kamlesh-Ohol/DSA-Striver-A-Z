```python
class Node:
    # Optimizes memory for every node created
    __slots__ = ['data', 'children']

    def __init__(self, data):
        self.data = data
        self.children = []  # Holds references to child Node objects

    def add_child(self, child_data):
        """Creates the child object inside the list dynamically."""
        new_node = Node(child_data)
        self.children.append(new_node)
        return new_node  # Returns the new child so you can add grandchildren to it

# ==========================================
# BUILDING THE TREE STARTING ONLY FROM ROOT
# ==========================================

# 1. Initialize just the root node
root = Node("A")

# 2. Add children directly through the root (returns the created node object)
node_b = root.add_child("B")
node_c = root.add_child("C")

# 3. Add a grandchild by calling add_child on its parent node
node_b.add_child("D")

# ==========================================
# VERIFYING THE STRUCTURE (Your print statements)
# ==========================================
print("Root:", root.data)

for child in root.children:
    print("Child:", child.data)

print("Child of B:", node_b.children[0].data)

# Printing the actual node objects/memory addresses in a clean list
for child in root.children:
    print(child.data)
```

Inorder Preorder Postorder

```python
class node:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None

def preorder(node,res):
    if node ==None:
       return 
    print(node.data)
    res.append(node.data)
    preorder(node.left,res)
    preorder(node.right,res)

def inorder(node,res):
    if node == None:
        return
    inorder(node.left,res)
    print(node.data)
    res.append(node.data)
    inorder(node.right,res)

def postorder(node,res):
    if node == None:
        return
    postorder(node.left,res)
    postorder(node.right,res)
    print(node.data)
    res.append(node.data)


root = node(1)
root.left = node(2)
root.right = node(3)
root.left.left = node(4)
root.left.right = node(5)
root.right.right = node(6)
root.right.left = node(7)
res = []
preorder(root,res)
print(res)
res.clear()
inorder(root,res)
print(res)
res.clear()
postorder(root,res)
print(res)

```

```python
class node:

  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None


def iterative_preorder(node, res):
  if node is None:
    return

  stack = [node]  # Initialize stack with the root node

  while stack:
    # Pop the top node from the stack
    current = stack.pop()

    # Process the node
    print(current.data)
    res.append(current.data)

    # Push right child first, then left child
    # (So left child is processed before right child due to LIFO)
    if current.right is not None:
      stack.append(current.right)
    if current.left is not None:
      stack.append(current.left)

def iterative_inorder(node, res):
  stack = []
  current = node
  while current is not None or stack:
    while current is not None:
      stack.append(current)
      current = current.left
    current = stack.pop()
    print(current.data)
    res.append(current.data)
    current = current.right



# Setting up the exact same tree as your example
root = node(1)
root.left = node(2)
root.right = node(3)
root.left.left = node(4)
root.left.right = node(5)
root.right.right = node(6)
root.right.left = node(7)

res = []
iterative_preorder(root, res)
res.clear()
print("$"*20)
iterative_inorder(root,res)
print(res)
```

Inorder Preorder Postorder in a single traversal

```python
def all_order_in_single_traversal(root):
    pre_order = []
    in_order = []
    post_order = []

    if root is None:
        return pre_order, in_order, post_order

    stack = [(root, 1)]

    while stack:
        node, state = stack.pop()

        # 1st time visiting the node
        if state == 1:
            pre_order.append(node.data)

            # Come back to this node for inorder
            stack.append((node, 2))

            # Traverse left subtree
            if node.left is not None:
                stack.append((node.left, 1))

        # 2nd time visiting the node
        elif state == 2:
            in_order.append(node.data)

            # Come back to this node for postorder
            stack.append((node, 3))

            # Traverse right subtree
            if node.right is not None:
                stack.append((node.right, 1))

        # 3rd time visiting the node
        else:
            post_order.append(node.data)

    return pre_order, in_order, post_order


class TreeNode:

  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
root.right.right = TreeNode(6)
root.right.left = TreeNode(7)
pre,in_,post = all_order_in_single_traversal(root)
print(pre)
print(in_)
print(post)
```

Check for Balanced Binary Tree

```python
class TreeNode:

  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None
def Balanced_Binary_Tree(node):
    if node ==None:
        return 0
    left_height = Balanced_Binary_Tree(node.left)
    if left_height == -1:
        return -1
    right_height = Balanced_Binary_Tree(node.right)
    if right_height == -1:
        return -1
    if abs(left_height-right_height)>1:
        return -1
    return max(left_height,right_height)+1

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
root.right.right = TreeNode(6)
root.right.left = TreeNode(7)
height = Balanced_Binary_Tree(root)
if height ==  0 or False:
   print("It's not a balanced Tree")
else:
   print("The height of a balanced Tree is ",height)
```

Diameter of a Tree

```python
class Node:
    def __init__(self,Data):
        self.Data = Data
        self.left = None
        self.right = None
        
def Diameter_of_a_tree(node,maxi):
    if node == None:
        return 0
    left_half = Diameter_of_a_tree(node.left,maxi)
    right_half = Diameter_of_a_tree(node.right,maxi)
    maxi[0] = max(maxi[0],left_half+right_half+1)
    return 1+max(left_half,right_half)

root = Node(1)
root.left = Node(2)
root.left.left = Node(4)
root.right = Node(3)
root.right.right = Node(5)
maxi = [0] 
Diameter_of_a_tree(root, maxi)

# Printing the total number of nodes along the longest path
print(f"Diameter of a tree (nodes) is {maxi[0]}") 
# Printing the number of edges along the longest path (if that is your target definition)
print(f"Diameter of a tree (edges) is {maxi[0] - 1}") 
```
