```python
class Stack:
    def __init__(self, data=None):
        self.stack = []
        if data is not None:
            self.stack.append(data)

    def push(self, data):
        if len(self.stack) >= 5:
            return "Stack Overflow"
        self.stack.append(data)
        return 1

    def pop(self):
        if len(self.stack) < 1:
            return "Stack Underflow"
        return self.stack.pop()

    def top(self):
        if len(self.stack) < 1:
            return "Stack is Empty"
        return self.stack[-1]

    def display(self):
        return self.stack


l1 = Stack(1)
l1.push(2)
l1.push(3)
l1.push(4)
l1.push(5)
print(l1.display())  
print(l1.push(6))    
print(l1.pop())
print(l1.pop())
print(l1.display())
```

```python

 
def Infix_to_Postfix(data):
    data = list(data)
    # print(data)
    order = {"+":1,"-":1,"*":2,"/":2,"^":3}
    i = 0
    ans = list()
    stack = list()
    size = len(data)
    while i!=size:
        if (data[i]>="A" and data[i]<="Z") or (data[i]>="a" and data[i]<="z") or (data[i]>="0" and data[i]<="9"):
            ans.append(data[i])
        elif data[i] =="(":
            stack.append(data[i])
        elif data[i]==")":
            while len(stack) ==0 and stack[-1]!="(":
                ans.append(stack.pop())
            stack.pop()
        else:
            while len(stack)!= 0 and order[data[i]]<= order[stack[-1]]:
                ans.append(stack.pop())
            stack.append(data[i])
        i+=1
        while(len(stack)!=0):
            ans = stack[-1]
            stack.pop()
    return ans


def Infix_to_Postfix(data):
    data = list(data)
    order = {"+": 1, "-": 1, "*": 2, "/": 2, "^": 3}
    i = 0
    ans = list()
    stack = list()
    size = len(data)
    
    while i != size:
        # 1. Operand check
        if (data[i] >= "A" and data[i] <= "Z") or (data[i] >= "a" and data[i] <= "z") or (data[i] >= "0" and data[i] <= "9"):
            ans.append(data[i])
            
        # 2. Opening parenthesis (must be elif, not if)
        elif data[i] == "(":
            stack.append(data[i])
            
        # 3. Closing parenthesis
        elif data[i] == ")":
            while len(stack) != 0 and stack[-1] != "(":
                ans.append(stack.pop())
            if len(stack) != 0:
                stack.pop()  # Remove '('
                
        # 4. Operators
        else:
            # Ensure stack[-1] != '(' before looking up precedence in order
            while len(stack) != 0 and stack[-1] != "(" and order[data[i]] <= order[stack[-1]]:
                ans.append(stack.pop())
            stack.append(data[i])
            
        i += 1

    # 5. Pop remaining operators OUTSIDE the main while loop
    while len(stack) != 0:
        ans.append(stack.pop())
        
    return "".join(ans)

print(Infix_to_Postfix("A+B*(C^D-E)"))



# Prefix_to_Postfix("A+B*(C^D-E)")
```
